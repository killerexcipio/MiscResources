```python
#File Finder

import csv
import os
import shutil

def extract_column_values(file_path, column_name):
    values = []

    # Open the CSV file
    with open(file_path, mode='r', newline='') as file:
        # Create a CSV reader object
        csv_reader = csv.DictReader(file)

        # Iterate over each row in the CSV
        for row in csv_reader:
            # Extract the value from the specified column and add it to the list
            value = row[column_name]
            values.append(value)

    return values

def copy_files_containing_values(source_folder, destination_folder, values_list):
    # Ensure destination folder exists
    os.makedirs(destination_folder, exist_ok=True)

    # Iterate over all files in the source folder
    for filename in os.listdir(source_folder):
        # Check if any value from the list is in the file name
        if any(value in filename for value in values_list):
            # Construct full file paths
            source_file = os.path.join(source_folder, filename)
            destination_file = os.path.join(destination_folder, filename)

            # Copy the file to the destination folder
            shutil.copy(source_file, destination_file)
            print(f"Copied: {filename} to {destination_folder}")

# Example usage (commented out to prevent execution)
csv_file_path = 'your_file.csv'
column_name = 'your_column_name'
result = extract_column_values(csv_file_path, column_name)
print(result)

source_folder = 'source_folder_path'
destination_folder = 'destination_folder_path'
copy_files_containing_values(source_folder, destination_folder, result)
```

```python
# File path
def cleaner(file_path):
  with open(file_path, 'r') as file:
      lines = file.readlines()
  if '    ' in lines[0]:
      lines.pop(0)
  cleaned_lines = []
  buffer = []

  for line in lines:
      stripped_line = line.strip()

      if stripped_line:
          buffer.append(stripped_line)
      else:
          if buffer:
              cleaned_lines.append(" ".join(buffer))
              buffer = []
          cleaned_lines.append("")
  # Add any remaining buffered lines
  if buffer:
      cleaned_lines.append(" ".join(buffer))
  cleaned_content = "\n".join(cleaned_lines)

  return cleaned_content
```

```python
#current final
import os
import csv

def check_letters(n):
    return any(char.isalpha() for char in n)

def transform_report(original_text):
    lines = original_text.splitlines(1)
    # if '    ' in lines[0]:
    #     lines.pop(0)  # Remove the first line if it contains redundant spaces

    clean_lines = []
    for line in lines:
        if ':' in line:
            _, post_colon = line.split(':', 1)
            clean_lines.append(post_colon.strip())
        else:
            clean_lines.append(line.strip())

    return clean_lines

def needs_quoting(text):
    # Check if text contains commas or multiple lines
    return ',' in text or text.count('.') > 1

def process_reports(input_folder, output_folder):
    os.makedirs(output_folder, exist_ok=True)

    for filename in os.listdir(input_folder):
        if filename.endswith('.txt'):
            input_path = os.path.join(input_folder, filename)
            # with open(input_path, 'r') as file:
            #     original_text = file.read()
            # print(original_text)
            original_text = cleaner(input_path)
            transformed_report = transform_report(original_text)

            output_filename = f"{os.path.splitext(filename)[0]}.csv"
            output_path = os.path.join(output_folder, output_filename)

            with open(output_path, mode='w', newline='') as file:
                writer = csv.writer(file, quoting=csv.QUOTE_MINIMAL)
                # Prepare the report for CSV output
                for i in transformed_report:
                  if i == '' or i == ' ':
                    continue
                  elif needs_quoting(i):
                      writer.writerow([f'"{i}"'])  # Write the report quoted if needed
                  else:
                      writer.writerow([i])  # Write the report without additional quotes

            print(f"Processed {filename} -> {output_filename}")

# Define your input and output folders
input_folder = '/content/P'
output_folder = '/content/trial'

process_reports(input_folder, output_folder)
```
